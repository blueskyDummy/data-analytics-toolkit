One correct way to solve each. Yours may differ and still be right, what matters is the output.

### 1. Variables and types

```
price = 42.50
print(type(price))
```

### 2. Arithmetic

```
price = 42.50
tip = price * 0.20
print(tip)
print(price + tip)
```

### 3. f-string

```
price = 42.50
total = price + price * 0.20
print(f"The total with tip is ${total}")
```

### 4. Comparison

```
print(42.50 > 50)
```

### 5. if / elif / else

```
score = 73
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("Below C")
```

### 6. for loop

```
scores = [88, 92, 70, 61, 95]
for s in scores:
    print(s)
```

### 7. Function

```
def total_with_tip(bill):
    return bill + bill * 0.18

print(total_with_tip(42.50))
```
