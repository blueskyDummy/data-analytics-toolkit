# Module 03 Solutions: NumPy

### 1.

```
import numpy as np
scores = np.array([88, 92, 70, 61, 95])
print(scores)
```

### 2.

```
print(scores + 5)
```

### 3.

```
print(scores.mean())
print(scores.max())
print(scores.min())
```

### 4.

```
print(scores[scores > 80])
```

### 5.

```
print((scores > 80).sum())
```
