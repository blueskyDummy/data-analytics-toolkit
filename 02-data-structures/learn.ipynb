# Module 02: Data Structures

Python has **four** built-in ways to hold a collection of values. This module covers all four, plus the one-line trick (comprehensions) for building them.

**By the end you will be able to:**
- use a **list** (ordered, changeable) and reach items by position
- use a **tuple** (ordered, fixed) and unpack it
- use a **dictionary** (look up by name)
- use a **set** (unique values, fast membership)
- build a new collection in one line with a **comprehension**

| Type | Looks like | Ordered? | Changeable? | Use it for |
|------|-----------|----------|-------------|------------|
| list | `[1, 2, 3]` | yes | yes | a sequence you'll edit |
| tuple | `(1, 2, 3)` | yes | **no** | a fixed record |
| dict | `{"a": 1}` | by key | yes | look up by name |
| set | `{1, 2, 3}` | no | yes | unique values |

Run every cell, then change a value and run again.

## 1. Lists: ordered and changeable
A **list** holds values in order, in square brackets `[ ]`. You can edit it after making it.

```
# a list of five restaurant bills
bills = [16.99, 10.34, 21.01, 23.68, 24.59]

print(bills)          # the whole list
print(len(bills))     # len() = how many items
```

### Indexing and slicing
Positions start at **0**. A slice `[start:stop]` stops *before* `stop`.

```
print(bills[0])     # first item  (position 0)
print(bills[-1])    # last item   (negative counts from the end)
print(bills[0:2])   # positions 0 and 1 -> first two
print(bills[2:])    # from position 2 to the end
```

### Lists can change
Add with `.append()`, or overwrite an item by its position.

```
bills.append(30.00)   # add a new bill at the end
bills[0] = 17.50      # overwrite the first item
print(bills)
```

## 2. Tuples: ordered but FIXED (immutable)
A **tuple** looks like a list but uses parentheses `( )`, and you **cannot change it** after creation. Use a tuple for a fixed record that shouldn't be edited, like a coordinate or a row of fields.

```
# one restaurant visit as a fixed record: (bill, tip, day)
visit = (16.99, 1.01, "Sunday")

print(visit)
print(visit[0])     # read by position, just like a list
print(len(visit))
```

Trying to change a tuple raises an error. We catch it here so the lesson is clear and the notebook keeps running.

```
try:
    visit[0] = 20.00          # attempt to change the first item
except TypeError as e:
    print("Tuples can't be changed:", e)
```

### Tuple unpacking (very handy)
You can split a tuple's values straight into separate variables in one line. You'll see this constantly, for example `df.shape` returns a tuple `(rows, cols)`.

```
bill, tip, day = visit        # unpack the 3 values into 3 names at once
print(bill)
print(tip)
print(day)

# real example: a DataFrame's .shape is a tuple you can unpack
shape = (244, 6)              # (rows, columns)
rows, cols = shape
print(f"{rows} rows and {cols} columns")
```

## 3. Dictionaries: look up by name
A **dictionary** stores `key: value` pairs in curly braces `{ }`. You look up a value by its key (a name) instead of a position. A pandas DataFrame is essentially a dict of columns.

```
# one visit described by field name
visit = {
    "bill": 16.99,     # key "bill" -> value 16.99
    "tip": 1.01,
    "day": "Sunday"
}

print(visit["bill"])   # look up the value stored under "bill"
print(visit["day"])
```

```
visit["people"] = 2       # add a new key:value pair
print(visit)

print(visit.keys())       # all the keys
print(visit.values())     # all the values

# loop over a dict: you get the keys
for field in visit:
    print(field, "->", visit[field])
```

## 4. Sets: unique values only
A **set** uses curly braces `{ }` like a dict but holds single values, not pairs. Two rules: every value is **unique** (duplicates vanish) and there is **no order**. Great for "what are the distinct values?" and fast "is X in here?" checks.

```
# the day of each of several visits (with repeats)
days = ["Sun", "Sun", "Sat", "Fri", "Sat", "Sun"]

unique_days = set(days)       # set() throws away the duplicates
print(unique_days)            # just the distinct days
print(len(unique_days))       # how many distinct days
```

### Fast membership and set math
`in` checks membership. Sets also do union `|`, intersection `&`, and difference `-`.

```
weekend = {"Sat", "Sun"}

print("Sat" in weekend)       # True  -> is Sat in the set?
print("Mon" in weekend)       # False

a = {"Sun", "Sat", "Fri"}
b = {"Sat", "Sun", "Mon"}
print(a & b)   # intersection: days in BOTH
print(a | b)   # union: days in EITHER
print(a - b)   # difference: in a but not b
```

## 5. Comprehensions: build a collection in one line
Instead of a loop that appends, describe the new list in one line: `[expression for item in collection]`. This is everywhere in data work.

```
bills = [16.99, 10.34, 21.01]

# the long way:
tips = []
for b in bills:
    tips.append(b * 0.18)
print(tips)

# the one-line comprehension (same result):
tips = [b * 0.18 for b in bills]       # "b*0.18 for each b in bills"
print(tips)
```

Add a condition with `if` at the end to keep only some items.

```
bills = [16.99, 10.34, 21.01, 8.50, 30.00]

big = [b for b in bills if b > 20]     # ...for each b... if it is > 20
print(big)
```

## Recap
Four built-in collections, each for a job:
- **list** `[ ]` — ordered, changeable; reach items by position
- **tuple** `( )` — ordered, fixed; a record you won't edit; unpack with `a, b = t`
- **dict** `{k: v}` — look things up by name
- **set** `{ }` — unique values, no order; dedupe and fast membership

And **comprehensions** build any of them in one readable line. These are the shapes almost all data takes.
