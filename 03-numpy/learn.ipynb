# Module 03: NumPy

**By the end you will be able to:**
- make a NumPy **array** and do math on all its values at once
- summarize an array (sum, mean, min, max, std)
- filter an array with a condition (boolean indexing)

NumPy is the engine under pandas. Its one big idea: do math on a **whole array at once**, with no loop. This is called *vectorization*, and it's both faster and easier to read.

## 1. Make an array
We import NumPy as `np` (the universal nickname). An array looks like a list but is built for math.

```
import numpy as np              # standard nickname

bills = np.array([16.99, 10.34, 21.01, 23.68])  # list -> array
print(bills)
print(type(bills))            # numpy.ndarray
```

## 2. Vectorized math: operate on everything at once
With a plain list you'd need a loop. With an array you just write the math once and it applies to every value.

```
# compute an 18% tip for EVERY bill in one line, no loop
tips = bills * 0.18           # multiplies each value by 0.18
print(tips)

totals = bills + tips         # adds the two arrays element by element
print(totals)
```

That `bills * 0.18` touching every element is the whole point of NumPy. Remember it from the regression and preprocessing code, this is why `df['income'] - mean` worked on a whole column.

## 3. Summarize an array
Arrays have built-in methods for the common statistics.

```
print(bills.sum())    # add them all up
print(bills.mean())   # average
print(bills.min())    # smallest
print(bills.max())    # largest
print(bills.std())    # standard deviation (spread)
```

## 4. Boolean indexing: filter by a condition
A comparison on an array gives an array of True/False. You can use that to keep only the values you want.

```
# which bills are over 20?
mask = bills > 20          # an array of True/False, one per value
print(mask)

print(bills[mask])         # keep only the values where mask is True
print(bills[bills > 20])   # the usual shortcut: do it in one step
```

This is exactly how row filtering works in pandas (`df[df['age'] > 30]`), same idea, one layer up.

## 5. A quick count with the condition
Summing a True/False array counts the Trues (True counts as 1).

```
print((bills > 20).sum())   # how many bills are over 20?
```

## Recap
A NumPy **array** lets you do math on all values at once (vectorization), summarize with `.mean()`, `.sum()`, etc., and filter with a boolean condition. Every one of these ideas reappears in pandas on columns.
