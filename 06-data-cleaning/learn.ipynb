# Module 06: Data Cleaning

**By the end you will be able to:**
- find and fill **missing values**
- remove **duplicate** rows
- fix **inconsistent** text labels

Real data is messy, and cleaning it is most of the job. We'll use `messy_people.csv`, which has every common problem baked in on purpose.

## 1. Look before you clean
Load the data and find the problems first. Never clean what you haven't inspected.

```
import pandas as pd

df = pd.read_csv('../data/messy_people.csv')
df
```

```
df.info()           # notice some columns have fewer non-null values -> missing data
```

## 2. Find missing values
`isna()` marks each cell True if it's missing. Summing counts the missing per column (True counts as 1).

```
df.isna()           # True wherever a value is missing
```

```
df.isna().sum()     # how many missing in each column
```

## 3. Fill missing values (imputation)
For a numeric column, a common fix is to fill blanks with the column's median (robust to outliers).

```
# fill missing age with the median age
median_age = df['age'].median()     # compute the middle value
df['age'] = df['age'].fillna(median_age)  # fillna replaces the blanks

# fill missing income with the median income
df['income'] = df['income'].fillna(df['income'].median())

df.isna().sum()     # check: should all be 0 now
```

## 4. Remove duplicate rows
`duplicated()` flags rows that are exact copies of an earlier row. `drop_duplicates()` removes them.

```
print(df.duplicated().sum())        # how many duplicate rows?

df = df.drop_duplicates()           # keep the first, drop the repeats
print(df.duplicated().sum())        # should be 0 now
```

## 5. Fix inconsistent text
The city column mixes 'Austin', 'austin', 'HOUSTON'. To the computer those are three different cities. Standardize the case so each city counts once.

```
print(df['city'].unique())          # see the inconsistent spellings

df['city'] = df['city'].str.title()  # .str.title() -> 'Austin', 'Houston'
print(df['city'].unique())          # now consistent
```

## 6. Handle an impossible value
One age is -5, which is impossible. Treat it as missing, then fill it like any other gap.

```
import numpy as np

df.loc[df['age'] < 0, 'age'] = np.nan      # mark impossible ages as missing
df['age'] = df['age'].fillna(df['age'].median())  # fill them

df
```

## 7. Save the cleaned data
Write the result to a NEW file, never overwrite the raw data.

```
df.to_csv('../data/_people_cleaned.csv', index=False)
print("saved cleaned file")
```

## Recap
The cleaning loop: inspect first, fill missing values (`isna`, `fillna`), drop duplicates (`drop_duplicates`), standardize messy text (`.str.title()`), convert impossible values to missing and fill them, then save to a new file. This is the exact workflow you'll run on your project data.
