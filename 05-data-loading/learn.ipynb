# Module 05: Data Loading

**By the end you will be able to:**
- read a normal CSV
- handle a CSV that has **no header row**
- read Excel and JSON
- save a DataFrame back out to a file

Real data arrives in many shapes. `read_csv` has options for the common surprises. The `data/` folder has small example files to practice on.

## 1. A normal CSV
`ex1.csv` has a header row (column names in the first line). This is the easy, common case.

```
import pandas as pd

df = pd.read_csv('../data/ex1.csv')   # first line is treated as column names
df
```

## 2. A CSV with NO header
`ex2.csv` has the same data but no names on the first line. If you read it normally, pandas wrongly uses the first data row as names. Two ways to fix it:

```
# option A: tell pandas there is no header, let it number the columns 0..4
pd.read_csv('../data/ex2.csv', header=None)
```

```
# option B: supply your own column names
pd.read_csv('../data/ex2.csv', names=['a','b','c','d','message'])
```

## 3. Choosing a column as the row index
`index_col` makes one column the row labels instead of a plain 0,1,2 index.

```
# use the 'message' column as the row index
pd.read_csv('../data/ex1.csv', index_col='message')
```

## 4. Excel files
`read_excel` works just like `read_csv`. (Reading .xlsx needs the `openpyxl` package, which is in our environment.)

```
xl = pd.read_excel('../data/ex1.xlsx')   # read the first sheet
xl
```

## 5. JSON
Data from web APIs often comes as JSON. `read_json` turns a simple JSON array into a DataFrame.

```
pd.read_json('../data/example.json')
```

## 6. Saving your work back out
Once you've built or cleaned a table, write it to a new file with `to_csv`. Use `index=False` so pandas doesn't add its row-number column.

```
df = pd.read_csv('../data/ex1.csv')
df.to_csv('../data/_my_output.csv', index=False)   # save without the index column
print("saved")
```

## Recap
`read_csv` handles the normal case; `header=None` or `names=` handles missing headers; `index_col=` sets the row labels; `read_excel` and `read_json` handle other formats; and `to_csv(..., index=False)` saves your result. Always keep the original file and write results to a new name.
